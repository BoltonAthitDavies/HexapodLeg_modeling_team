```python
import pandas as pd
print()
```

```python
import os
path = '/home/ambushee/motor_logs/'
table = os.listdir(path)
print(table)
```

```python
name = table[1]
print(name)
df = pd.read_csv(path + name)
df.head()
```

```python
# simple plot
fig = plt.figure(figsize=(30, 20))
plt.title('Motor 1 Acceleration over Time', fontsize=40 )
plt.legend(fontsize=16)
plt.grid(True, alpha=0.3)
plt.xlabel('time_sec', fontsize=40)
plt.ylabel('rad / s', fontsize=40)
plt.plot(df['time_sec'], df['motor_1_acceleration'])

plt.tight_layout()
plt.show()
```

```python
# simple plot
fig = plt.figure(figsize=(30, 20))
plt.title('Motor 1 Effort over Time', fontsize=40 )
plt.legend(fontsize=16)
plt.grid(True, alpha=0.3)
plt.xlabel('time_sec', fontsize=40)
plt.ylabel('n.m', fontsize=40)
plt.plot(df['time_sec'], df['motor_1_effort'])

plt.tight_layout()
plt.show()
```
